# Research Question

Our research question is: using questionnaire data, to what extent are chronotype and sleep quality factors associated with GPA among college students? This study will ask students to self-report information pertaining to sleep habits, demographic factors, and other lifestyle variables that are relevant in assessing sleep and chronotype factors’ correlations with academic performance. This study will use inferential statistics, mainly multiple regression and ANOVA tests, to assess the proportion of variance in GPA that directly correlates with sleep quality and chronotype. (Variables used are discussed more in-depth in the ‘Data’ section)

## Background and Prior Work

Sleep has been shown to play a critical role in cognitive functioning, memory consolidation, and learning, all of which are essential for academic success. Prior research has consistently shown that sleep deprivation is associated with lower academic performance (GPA) among college students (Hershner & Chervin, 2014). Incorporating this into our design, we recognize that sleep is a multidimensional construct that includes more than just sleep duration as a significant predictor of overall sleep quality. As a result, this study aims to reinforce the literature on other sleep subvariables’ associations with academic performance among college students. Along with the quantity of sleep, the continuity, variability, and timing of sleep periods are also essential factors in assessing sleep health (Buysse, 2014). 

Related to sleep timing is chronotype, which refers to an individual’s natural preference for activity at certain times of day influenced by genetic and environmental factors. Oftentimes, there exists a misalignment between inflexible scheduled activities, such as work and school, and natural chronotype preferences. This mismatch between natural chronotype and work/school schedules is referred to as “social jetlag” (Wittmann et al., 2009). We represent this construct as the difference between the sleep midpoint on free days versus work/school days. The presence of social jetlag has been linked to poorer academic outcomes and reduced wellbeing, as reduced sleep quality on working days is compensated for by increased sleep duration on non-working days, thus creating harmful inconsistencies in long-term sleep schedule. 

This study aims to combine multidimensional measures of sleep quality with chronotype category to assess the extent to which each variable predicts academic performance. In order to better isolate these variables from potential confounds, caffeine consumption and phone usage before bed will be recorded from participants, as both of these effects have known effects on delaying circadian rhythms and interfering with sleep quality measures. Demographic variables, particularly age and sex, are also related to chronotype expression (Roennenberg, 2012). Our model of chronotype and sleep quality will account for these influencing variables to ensure accuracy in detecting an independent effect of chronotype on GPA.


# Hypothesis


We hypothesize that sleep behavior and chronotype alignment will significantly predict academic performance (GPA) among college students. "Perceived chronotype" refers to an individual's subjective experience of their internal biological clock. It dictates natural preferences for sleep and wakefulness. In our study, we predict that students who report 7-9 hours of sleep per night will have a higher GPA compared to those with shorter or longer sleep duration. We further hypothesize that greater chronotype mismatch, defined as the discrepancy between preferred sleep timing and actual daily schedules, will be associated with lower GPA. Additionally, we predict that later sleep onset times will negatively relate to GPA due misalignment with typical academic schedules. Finally, we hypothesize that these effects will be moderated by academic context, with stronger negative impacts observed among STEM majors and upper-division students. 

We predict that students who experience greater mismatch between their preferred sleep schedules and their actual weekday sleep schedules, along with lower sleep consistency and poorer overall sleep habits, will tend to report lower GPAs. In contrast, students whose sleep patterns are more aligned with their weekday academic schedules are expected to report stronger academic performance overall.
