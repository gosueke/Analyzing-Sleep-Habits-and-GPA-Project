## Abstract

Sleep is widely recognized as an important factor in cognitive performance, memory consolidation, and overall academic success. In this project, we are investigating the relationship between sleep habits, but specifically chronotype alignment, sleep quality, and behavioral sleep habits as a means to determine the success of undergraduate college students. We measure success through self reported GPA among the undergraduate students. In our project, we are particularly interested in whether misalignment between preferred sleep timing (chronotype) and actual sleep schedules, as well as other sleep related behaviors, are associated with differences in students’ GPA. Our hypothesis was that students with consistent sleep habits that are getting about 7-9 hours of sleep will have greater alignment between their preferred and actual sleep schedules and will have higher GPAs, whereas those with a greater chronotype mismatch, poorer sleep habits, and later sleep times will be associated with lower academic performance (lower GPA).

In order to explore these relationships, we analyzed survey data collected from 134 college undergraduate students using a structured questionnaire covering sleep timing, sleep duration, sleep quality indicators, lifestyle behaviors, and demographic variables. After data cleaning and filtering, the final analytic sample consisted of 49 participants, and from that we conducted multiple linear regression analyses, quadratic models, correlation analysis, and ANOVA tests to evaluate how sleep related variables predict GPA. We also controlled for potential confounds such as sex, major category, caffeine consumption, and phone usage before bedtime.

Our results show that chronotype mismatch and sleep timing variables have weak or non significant associations with GPA. However, one consistent finding emerged increased pre bed phone usage is significantly associated with lower GPA, even after accounting for other sleep related variables. This relationship remained robust in both linear and quadratic models and represented the strongest predictor of GPA in our dataset.

Overall, our findings provide very limited support for our original hypothesis. While we expected that greater chronotype alignment, consistent sleep habits, and obtaining 7–9 hours of sleep would be associated with higher GPAs, we found little evidence that sleep duration or chronotype mismatch significantly predicted academic performance in our sample. Instead, increased phone usage before bedtime emerged as the strongest predictor of lower GPA, which suggests that specific behavioral sleep habits may have a greater influence on academic performance than chronotype alignment alone.

## Background and Prior Work

Sleep has been shown to play a critical role in cognitive functioning, memory consolidation, and learning, all of which are essential for academic success. Prior research has consistently shown that sleep deprivation is associated with lower academic performance (GPA) among college students (Hershner & Chervin, 2014). Incorporating this into our design, we recognize that sleep is a multidimensional construct that includes more than just sleep duration as a significant predictor of overall sleep quality. As a result, this study aims to reinforce the literature on other sleep subvariables’ associations with academic performance among college students. Along with the quantity of sleep, the continuity, variability, and timing of sleep periods are also essential factors in assessing sleep health (Buysse, 2014). 

Related to sleep timing is chronotype, which refers to an individual’s natural preference for activity at certain times of day influenced by genetic and environmental factors. Oftentimes, there exists a misalignment between inflexible scheduled activities, such as work and school, and natural chronotype preferences. This mismatch between natural chronotype and work/school schedules is referred to as “social jetlag” (Wittmann et al., 2009). We represent this construct as the difference between the sleep midpoint on free days versus work/school days. The presence of social jetlag has been linked to poorer academic outcomes and reduced wellbeing, as reduced sleep quality on working days is compensated for by increased sleep duration on non-working days, thus creating harmful inconsistencies in long-term sleep schedule. 

This study aims to combine multidimensional measures of sleep quality with chronotype category to assess the extent to which each variable predicts academic performance. In order to better isolate these variables from potential confounds, caffeine consumption and phone usage before bed will be recorded from participants, as both of these effects have known effects on delaying circadian rhythms and interfering with sleep quality measures. Demographic variables, particularly age and sex, are also related to chronotype expression (Roennenberg, 2012). Our model of chronotype and sleep quality will account for these influencing variables to ensure accuracy in detecting an independent effect of chronotype on GPA.



## Hypothesis


We hypothesize that sleep behavior and chronotype alignment will significantly predict academic performance (GPA) among college students. "Perceived chronotype" refers to an individual's subjective experience of their internal biological clock. It dictates natural preferences for sleep and wakefulness. In our study, we predict that students who report 7-9 hours of sleep per night will have a higher GPA compared to those with shorter or longer sleep duration. We further hypothesize that greater chronotype mismatch, defined as the discrepancy between preferred sleep timing and actual daily schedules, will be associated with lower GPA. Additionally, we predict that later sleep onset times will negatively relate to GPA due misalignment with typical academic schedules. Finally, we hypothesize that these effects will be moderated by academic context, with stronger negative impacts observed among STEM majors and upper-division students.

We predict that students who experience greater mismatch between their preferred sleep schedules and their actual weekday sleep schedules, along with lower sleep consistency and poorer overall sleep habits, will tend to report lower GPAs. In contrast, students whose sleep patterns are more aligned with their weekday academic schedules are expected to report stronger academic performance overall.

### Exploring GPA Relationships with Chronotype and Sleep Factors

# Regression of GPA and Chronotype Mismatch

To begin we will look at GPA as a function of Chronotype mismatch. This explores whether grade point average varies depending on chronotype mismatch. We predicted that higher chronotype mismatch would lead to a lower GPA. Let's see!

```python
#regression model predicting GPA from chronotype mismatch
chrono_mismatch = smf.ols('gpa ~ chronotype_mismatch',data=df).fit()
print(chrono_mismatch.summary())

#remove outliers for chronotype mismatch
df.loc[39,'free_day_bedtime'] -= 12
df.loc[39,'free_day_sleep_duration'] -= 12
df.loc[39,'free_day_sleep_midpoint'] = 4.75
df.loc[39,'chronotype_mismatch'] = 0
df.loc[39]

#plotting chronotype mismatch and GPA
sns.regplot(data=df,x='chronotype_mismatch',y='gpa')
plt.title('Student GPA by Chronotype Mismatch')
plt.xlabel('Chronotype Mismatch')
plt.ylabel('GPA')
plt.show()
```

Unfortunately, this produced a very small R-squared value of 0.011. Next, we'll see if sleep midpoints can act as a predictor for GPA. We hypothesize that later midpoints will lead to a lower GPA. 

```python
#checking for multicolinearity-- the correlation is high, so we'll just use the weekday sleep midpoint in the model
print(df[['weekday_sleep_midpoint','free_day_sleep_midpoint']].corr())

#regression model predicting GPA from weekday sleep midpoint
chrono_midpoint = smf.ols('gpa ~ weekday_sleep_midpoint',data=df).fit()
print(chrono_midpoint.summary())

sns.regplot(data=df,x='weekday_sleep_midpoint',y='gpa')
plt.title('Student GPA by Weekday Sleep Midpoint')
plt.xlabel('Weekday Sleep Midpoint')
plt.ylabel('GPA')
plt.show()
```

There is a slightly stronger relationship here, with an R-squared value of 0.049. This is still FAR from significant, however, and certainly influenced by the outlier. Lastly, we are going to see if there is a linear relationship between well-restedness and GPA.

```python
#regression model predicting GPA from well restedness
well_restedness = smf.ols('gpa ~ well_rested',data=df).fit()
print(well_restedness.summary())

sns.regplot(data=df,x='well_rested',y='gpa')
plt.title('Student GPA by Well Restedness')
plt.xlabel('Well Restedness')
plt.ylabel('GPA')
plt.show()
```

 This produced a small R-squared value of 0.006, so we found no relationship between well-restedness and GPA. As a check, we are going to do an exploratory ANOVA using categorical predictors collected from our survey, to ensure these aren't influencing GPA (as opposed to the variables we want to manipulate). 

```python
sex_model = smf.ols('gpa ~ C(sex)',data=df).fit()
print(sm.stats.anova_lm(sex_model))

plt.figure(figsize=(6,5))
sns.boxplot(data=df,x='sex',y='gpa')
plt.title('GPA by Sex')
plt.xlabel('Sex')
plt.ylabel('GPA')
plt.show()
```

As expected, sex is not a good predictor of GPA. Major type could be a better predictor, as it is plausible that some majors are harder than others.

```python

major_model = smf.ols('gpa ~ C(major_category)',data=df).fit()
print(sm.stats.anova_lm(major_model))

plt.figure(figsize=(8,5))
sns.boxplot(data=df,x='major_category',y='gpa')
plt.title('GPA by Major Category')
plt.xlabel('Major Category')
plt.ylabel('GPA')
plt.xticks(rotation=20)
plt.show()
```

While the Social Sciences seems to be a bit higher than Arts and Humanities, there is no significant relationship between major and GPA. Last, we'll make a model to see if the frequency of night awakenings affects GPA.

```python
wake_model = smf.ols('gpa ~ C(night_wake_frequency)',data=df).fit()
print(sm.stats.anova_lm(wake_model))



plt.figure(figsize=(8,5))
sns.boxplot(data=df,x='night_wake_frequency',y='gpa')
plt.title('GPA by Night Wake Frequency')
plt.xlabel('Night Wake Frequency')
plt.ylabel('GPA')
plt.xticks(rotation=20)
plt.show()
```

Even though night wake frequency impedes sleep quality, this plot shows that it is not having a significant effect on GPA, which is good for the quality of our analysis. Overall, sex, major, and night wake frequency all show non-significant relationship with GPA. This means that any relationship we see is the result of the variable we are testing, and not one of these three confounds. 

### Exploring Nonlinear Analysis and Other Predictors

So far, none of our linear models have produced significant relationships. This is likely because our sample size is too small to observe an effect. However, it is also worthy to note that the quality of academic performance increases exponentially as GPA gets higher. Most of our data points were between 3.0 and 4.0, and it is meaningful to distinguish the difference between a 3.8-4.0 and a 3.1-3.3. For this reason we will now see what these relationships look like using a quadratic model, and explore some other variables we collected data on in our survey.

```python
#creating a quadratic model for weekday sleep duration
week_sleep_duration = smf.ols('gpa ~ weekday_sleep_duration + I(weekday_sleep_duration**2)',data=df).fit()
print(week_sleep_duration.summary())

sns.regplot(data=df,x='weekday_sleep_duration',y='gpa',order=2)
plt.title('Student GPA by Weekday Sleep Duration')
plt.xlabel('Weekday Sleep Duration')
plt.ylabel('GPA')
plt.show()
```

This non-linear model has an adjusted R-squared value of -0.042, which is far from significant, and even less so than our linear model. So, we'll now create a correlation heatmap to see other sleep variables' relationships with each other and GPA to explore which factors may explain variance in GPA.

```python

corr_vars = [
    'age',
    'well_rested',
    'sleep_consistency',
    'sleep_latency_minutes',
    'pre_bed_phone_minutes',
    'late_caffeine_days',
    'gpa'
]

corr_df = df[corr_vars].apply(pd.to_numeric, errors='coerce')

corr_matrix = corr_df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```

There appears to be a potentially meaningful relationship between GPA, sleep latency, pre-bed phone minutes, and late caffeine days. We'll construct and evaluate a regression model from the three strongest predictors. 

```python

latency_model = smf.ols('gpa ~ sleep_latency_minutes + pre_bed_phone_minutes + late_caffeine_days',data=df).fit()
print(latency_model.summary())
```

Of these three predictor variables, phone usage before bed is the only significant predictor of GPA, with a p-value of 0.033. So, let's make try a linear model only including phone usage.


```python
phone_model = smf.ols('gpa ~ pre_bed_phone_minutes',data=df).fit()
print(phone_model.summary())

sns.regplot(data=df,x='pre_bed_phone_minutes',y='gpa')
plt.title('Student GPA by Pre-Bed Phone Usage')
plt.xlabel('Minutes of Phone Usage')
plt.ylabel('GPA')
plt.show()
```

There seems to be a strong relationship between minutes of phone usage before bed and GPA! This model produced an R-squared value of 0.12, which is our strongest thus far! Let's use this predictor to construct a quadratic model.

```python

#creating a quadratic model for weekday sleep duration


quad_phone_model = smf.ols(
    'gpa ~ pre_bed_phone_minutes + I(pre_bed_phone_minutes**2)',
    data=df
).fit()
print(quad_phone_model.summary())

sns.regplot(data=df,x='pre_bed_phone_minutes',y='gpa',order=2)
plt.title('Student GPA by Pre-Bed Phone Usage')
plt.xlabel('Pre-Bed Phone Minutes')
plt.ylabel('GPA')
plt.show()
```

This gave an even stronger R-squared value of 0.134! Given our low sample size, this is not enough to make any significant claims, but this finding does align with knowledge about the negative impacts of blue light on circadian rhythm. 

### What does this mean?

Overall, our strongest finding was that higher pre-bed phone minutes are correlated with a lower GPA. This relationship strengthened when using a nonlinear model, which suggests that GPA is better represented by a quadratic slope. This makes sense, as it is much easier to lower a GPA than to raise it. Differences between 3.8 and 4.0 reflect greater discrepencies than differences between 2.0 and 2.2. 

 Given our results, we can't advise you to go to sleep earlier, or sleep for longer. But, it seems like putting your phone away before bed may be a good idea.

## Discussion and Conclusion

This study examined the relationship between chronotype, sleep quality, and academic performance among college students. Based on previous research, we expected that greater chronotype mismatch and poorer sleep habits would be associated with lower GPAs. Our findings only partially supported this hypothesis. While chronotype mismatch showed little relationship with GPA, other sleep-related behaviors appeared to have stronger associations with academic performance. In particular, pre-bed phone usage demonstrated the strongest negative relationship with GPA among the variables examined.

These findings build upon previous literature that has identified sleep as an important factor in learning, memory consolidation, and academic success. Earlier research suggested that disruptions in sleep schedules and social jetlag may negatively impact academic performance. Although our study did not find strong evidence that chronotype mismatch itself predicts GPA, our results suggest that behaviors surrounding sleep may be more important than sleep timing alone. Students who reported spending more time on their phones before bed generally reported lower GPAs, indicating that nighttime habits may influence academic outcomes through their effects on sleep quality and overall rest. 

Taken together, our findings suggest that academic performance may be influenced by a combination of lifestyle behaviors rather than any single sleep measure. While sleep duration, chronotype mismatch, sleep latency, late caffeine use, sex, major category, and night wake frequency did not significantly predict GPA in our analysis, pre-bed phone usage showed a meaningful negative relationship with academic performance. This may indicate that habits that interfere with healthy sleep routines have a greater impact on academic success than previously expected. These results highlight the complexity of sleep and suggest that multiple factors contribute to academic performance among college students.

Some of our results may not have fully matched our original hypotheses. There are several possible explanations for this. Academic performance is influenced by many factors beyond sleep, including study habits, workload, stress, motivation, extracurricular involvement, employment responsibilities, and personal circumstances. Because these variables can vary widely across students, they may have reduced or obscured the strength of some relationships observed in our analysis.

There are several limitations that should be considered when interpreting these findings. First, our data relied on self-reported survey responses, which may introduce recall bias or reporting inaccuracies. Second, participation was voluntary, which may have resulted in sampling bias and reduced the representativeness of the sample. Additionally, GPA was self-reported rather than obtained from official records, and other potentially important variables such as mental health, socioeconomic status, and time-management skills were not directly measured.

From an ethical perspective, it is important not to overgeneralize our results or imply causal relationships. Our study was observational and cross-sectional, meaning that we can identify associations but cannot determine whether sleep behaviors directly cause changes in GPA. While we took steps to minimize bias, protect participant privacy, and collect data responsibly, the findings should be interpreted as correlations rather than definitive explanations of academic performance. Sleep is only one of many factors that contribute to student success.

Future research could strengthen these findings by collecting larger and more diverse samples, incorporating longitudinal data, and using objective measures of sleep such as sleep tracking devices. Additional studies could also examine how factors such as major, academic level, employment status, stress, and mental health interact with sleep behaviors. Overall, this project contributes to the growing body of research on sleep and academic achievement and highlights the importance of considering chronotype and sleep quality when evaluating factors associated with college student success.
