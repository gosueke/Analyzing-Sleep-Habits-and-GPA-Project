## Research Question

Our research question is: using questionnaire data, to what extent are chronotype and sleep quality factors associated with GPA among college students? This study will ask students to self-report information pertaining to sleep habits, demographic factors, and other lifestyle variables that are relevant in assessing sleep and chronotype factors’ correlations with academic performance. This study will use inferential statistics, mainly multiple regression and ANOVA tests, to assess the proportion of variance in GPA that directly correlates with sleep quality and chronotype. (Variables used are discussed more in-depth in the ‘Data’ section)

## Background and Prior Work

Sleep has been shown to play a critical role in cognitive functioning, memory consolidation, and learning, all of which are essential for academic success. Prior research has consistently shown that sleep deprivation is associated with lower academic performance (GPA) among college students<a name="cite_ref-1"></a><sup>1</sup>
. However, sleep is a multidimensional construct that includes more than just sleep duration as a significant predictor of overall sleep quality. This study aims to reinforce the literature on other sleep subvariables’ associations with academic performance among college students. Along with the quantity of sleep, the continuity, variability, and timing of sleep periods are also essential factors in assessing sleep health<a name="cite_ref-2"></a><sup>2</sup>
.

Related to sleep timing is chronotype, which refers to an individual’s natural preference for activity at certain times of day influenced by genetic and environmental factors. Oftentimes, there exists a misalignment between inflexible scheduled activities, such as work and school, and natural chronotype preferences. This mismatch between natural chronotype and work/school schedules is referred to as “social jetlag”<a name="cite_ref-3"></a><sup>3</sup>
. The presence of social jetlag has been linked to poorer academic outcomes and reduced wellbeing, as reduced sleep quality on working days is compensated for by increased sleep duration on non-working days, thus creating harmful inconsistencies in long-term sleep schedule.

This study aims to combine multidimensional measures of sleep quality with chronotype category to assess the extent to which each variable predicts academic performance. In order to better isolate these variables from potential confounds, caffeine consumption and phone usage before bed will be recorded from participants, as both of these factors have known effects on delaying circadian rhythms and interfering with sleep quality measures. Demographic variables, particularly age and sex, are also related to chronotype expression<a name="cite_ref-4"></a><sup>4</sup>
. Our model of chronotype and sleep quality will account for these influencing variables to ensure accuracy in detecting an independent effect of chronotype on GPA.

## Hypothesis


We hypothesize that sleep behavior and chronotype alignment will significantly predict academic performance (GPA) among college students. Specifically, we predict that students who report 7-9 hours of sleep per night will have a higher GPA compared to those with shorter or longer sleep duration. We further hypothesize that greater chronotype mismatch, defined as the discrepancy between preferred sleep timing and actual daily schedules, will be associated with lower GPA. Additionally, we predict that later sleep onset times will negatively relate to GPA due misalignment with typical academic schedules. Finally, we hypothesize that these effects will be moderated by academic context, with stronger negative impacts observed among STEM majors and upper-division students.

The dataset for this project would include self-reported data on sleep behavior, academic performance, and contextual academic variables. Key variables would include: average sleep duration (hours per night), sleep onset time, wake time, preferred study time (chronotype), actual study time, GPA (self-reported), major type (categorized as STEM, Humanities, Social Sciences, Business, other), and proportion of upper-division coursework. Additional covariates such as year in school, work hours, and caffeine consumption will be included to control for potential confounding effects. 
With GPA as a dependent variable and sleep length, chronotype mismatch, and sleep time as main predictors, these factors enable us to evaluate our hypothesis using multiple regression models. The disparity between desired and actual schedules will be used to operationalize chronotype mismatch. To investigate whether academic context (e.g., STEM vs. non-STEM, upper-division status) influences these relationships, interaction terms will be included. 
Undergraduate students will be given an online Google form to complete in order to gather data. Given the amount of factors in our regression model, we want to gather at 150-200 answers to guarantee enough statistical power. Anonymized response will be safely kept in a Google drive password- protected folder. 
This study uses self-reported survey data, which may introduce recall and reporting bias, whereas the ideal dataset would contain longitudinal or objectively measured sleep data (e.g., actigraphy or wearable device data). We will investigate publicly accessible datasets, such as the National College Health Assessment (NCHA) or other extensive student health surveys, which contain sleep and academic performance factors, in addition to the data we have gathered. These datasets could serve as a benchmark or additional confirmation of our conclusions.
