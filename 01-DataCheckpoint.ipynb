## Research Question

Our research question is: using questionnaire data, to what extent are chronotype and sleep quality factors associated with GPA among college students? This study will ask students to self-report information pertaining to sleep habits, demographic factors, and other lifestyle variables that are relevant in assessing sleep and chronotype factors’ correlations with academic performance. This study will use inferential statistics, mainly multiple regression and ANOVA tests, to assess the proportion of variance in GPA that directly correlates with sleep quality and chronotype. (Variables used are discussed more in-depth in the ‘Data’ section)

## Background and Prior Work

Sleep has been shown to play a critical role in cognitive functioning, memory consolidation, and learning, all of which are essential for academic success. Prior research has consistently shown that sleep deprivation is associated with lower academic performance (GPA) among college students (Hershner & Chervin, 2014). Incorporating this into our design, we recognize that sleep is a multidimensional construct that includes more than just sleep duration as a significant predictor of overall sleep quality. As a result, this study aims to reinforce the literature on other sleep subvariables’ associations with academic performance among college students. Along with the quantity of sleep, the continuity, variability, and timing of sleep periods are also essential factors in assessing sleep health (Buysse, 2014). 

Related to sleep timing is chronotype, which refers to an individual’s natural preference for activity at certain times of day influenced by genetic and environmental factors. Oftentimes, there exists a misalignment between inflexible scheduled activities, such as work and school, and natural chronotype preferences. This mismatch between natural chronotype and work/school schedules is referred to as “social jetlag” (Wittmann et al., 2009). We represent this construct as the difference between the sleep midpoint on free days versus work/school days. The presence of social jetlag has been linked to poorer academic outcomes and reduced wellbeing, as reduced sleep quality on working days is compensated for by increased sleep duration on non-working days, thus creating harmful inconsistencies in long-term sleep schedule. 

This study aims to combine multidimensional measures of sleep quality with chronotype category to assess the extent to which each variable predicts academic performance. In order to better isolate these variables from potential confounds, caffeine consumption and phone usage before bed will be recorded from participants, as both of these effects have known effects on delaying circadian rhythms and interfering with sleep quality measures. Demographic variables, particularly age and sex, are also related to chronotype expression (Roennenberg, 2012). Our model of chronotype and sleep quality will account for these influencing variables to ensure accuracy in detecting an independent effect of chronotype on GPA.


## Hypothesis


We hypothesize that sleep behavior and chronotype alignment will significantly predict academic performance (GPA) among college students. "Perceived chronotype" refers to an individual's subjective experience of their internal biological clock. It dictates natural preferences for sleep and wakefulness. In our study, we predict that students who report 7-9 hours of sleep per night will have a higher GPA compared to those with shorter or longer sleep duration. We further hypothesize that greater chronotype mismatch, defined as the discrepancy between preferred sleep timing and actual daily schedules, will be associated with lower GPA. Additionally, we predict that later sleep onset times will negatively relate to GPA due misalignment with typical academic schedules. Finally, we hypothesize that these effects will be moderated by academic context, with stronger negative impacts observed among STEM majors and upper-division students. 

We predict that students who experience greater mismatch between their preferred sleep schedules and their actual weekday sleep schedules, along with lower sleep consistency and poorer overall sleep habits, will tend to report lower GPAs. In contrast, students whose sleep patterns are more aligned with their weekday academic schedules are expected to report stronger academic performance overall.

```python
#Tidy Data
# replace the titles that qualtrics automatically generated

raw_file.columns = raw_file.iloc[0]
new_header = raw_file[2:].reset_index(drop=True)
new_header.head()
```

```python
print(new_header.columns)
```

```python
#Clean Data

#drops the data that qualtrics automatically generated with the survey as well as the answers with 'Nan' responses
#Using dropna() was also the best choice in case of our data because it ensures that only participants who provided 
#complete responses are included in your analysis. Because many of our variables like bedtime, wake time, sleep latency, 
#caffeine use, GPA, and extracurricular hours are  being compared against each other, missing values would make
#statistical analysis inconsistent and harder to interpret.
#These NA values appear to be random errors in participant responses or lack of full participation.

drop_unnecessary = new_header.drop(columns = 
                [ 'Start Date', 'End Date', 'Response Type', 'IP Address', 'Progress',
                  'Duration (in seconds)', 'Finished', 'Recorded Date',
                  'Recipient Last Name', 'Recipient First Name', 'Recipient Email',
                  'External Data Reference', 'Location Latitude', 'Location Longitude',
                  'Distribution Channel', 'User Language' ]).dropna()
drop_unnecessary
```

```python
#Clean Data

#Renames the columns for better readability and removes columns only useful for participant screening. Shows the shape of the dataframe after removing NAs and irrelevant columns
#Note that all data is measure in the Past week and all Scales are weighed from 0-5 with 0 being not
#rested and 5 being well rested (Rested Score, and Sleep Consistency)

rename_columns = drop_unnecessary.rename(columns=
               {
                 drop_unnecessary.columns[1]: 'consent',
                 drop_unnecessary.columns[2]: 'sleep_disorder',
                 drop_unnecessary.columns[3]: 'eligible_undergrad',
                 drop_unnecessary.columns[4]: 'sex',
                 drop_unnecessary.columns[5]: 'age',
                 drop_unnecessary.columns[6]: 'weekday_bedtime',
                 drop_unnecessary.columns[7]: 'weekday_wakeup',
                 drop_unnecessary.columns[8]: 'free_day_bedtime',
                 drop_unnecessary.columns[9]: 'free_day_wakeup',
                 drop_unnecessary.columns[10]: 'well_rested',
                 drop_unnecessary.columns[11]: 'sleep_consistency',
                 drop_unnecessary.columns[12]: 'night_wake_frequency',
                 drop_unnecessary.columns[13]: 'sleep_latency_minutes',
                 drop_unnecessary.columns[14]: 'pre_bed_phone_minutes',
                 drop_unnecessary.columns[15]: 'late_caffeine_days',
                 drop_unnecessary.columns[16]: 'gpa',
                 drop_unnecessary.columns[17]: 'major_category',
                 drop_unnecessary.columns[18]: 'extracurricular_hours_per_day',                                                       
                                                                        })
rename_columns = rename_columns.drop(columns = ['consent','sleep_disorder','eligible_undergrad'])

print(rename_columns.shape)
rename_columns.head()
```

```python
#Clean Data

#Sets Index to be responder Id so that it is unique for each responder. 

unique_index = rename_columns.iloc[1:].reset_index(drop=True)
unique_index = unique_index.set_index("Response ID")
unique_index.head()
```

```python
#Converts Numeric data to numeric data type

numeric_cols = [
    'age',
    'weekday_bedtime',
    'free_day_bedtime',
    'weekday_wakeup',
    'free_day_wakeup',
    'sleep_latency_minutes',
    'pre_bed_phone_minutes',
    'well_rested',
    'sleep_consistency',
    'late_caffeine_days',
    'gpa',
    'extracurricular_hours_per_day'
]

clean_num_cat = unique_index.copy()
clean_num_cat[numeric_cols] = clean_num_cat[numeric_cols].apply(pd.to_numeric,errors='coerce')

clean_num_cat.head()
```

```python
#Clean Data
# Checks that GPA <= 4.0, and there is no negative sleep latency and removes data that doesn't fall within the boundary

real_values = clean_num_cat[
    (clean_num_cat['gpa'].between(0, 4.0)) &
    (clean_num_cat['sleep_latency_minutes'] >= 0) 
]
real_values
```

```python
#We asked participants to report their sleep and wake times as 24hr time scale numbers, but only some complied
#This code checks if a participants bedtime is reported as 8am-12pm (really strange) and checks that they don't use 24hr scale for the wake time
#for datapoints that pass this check, 12 is added to their value to correct them

error1 = (
    real_values['weekday_bedtime'].between(8,12) &
    ~real_values['weekday_wakeup'].between(13,24)
)
error2 = (
    real_values['free_day_bedtime'].between(8,12) &
    ~real_values['free_day_wakeup'].between(13,24)
)

corrected_values = real_values.copy()

corrected_values.loc[error1, 'weekday_bedtime'] += 12
corrected_values.loc[error2, 'free_day_bedtime'] += 12

corrected_values.head()
```

```python
#From the clean data we can now calculate chronotype through sleep midpoint, sleep duration, and chronotype mismatch between midpoints


corrected_values['weekday_sleep_duration'] = (
    corrected_values['weekday_wakeup'] -
    corrected_values['weekday_bedtime']) % 24

corrected_values['free_day_sleep_duration'] = (
    corrected_values['free_day_wakeup'] -
    corrected_values['free_day_bedtime']) % 24

corrected_values['weekday_sleep_midpoint'] = (
    corrected_values['weekday_bedtime'] +
    corrected_values['weekday_sleep_duration'] / 2) % 24

corrected_values['free_day_sleep_midpoint'] = (
    corrected_values['free_day_bedtime'] +
    corrected_values['free_day_sleep_duration'] / 2) % 24

corrected_values['chronotype_mismatch'] = (
    corrected_values['free_day_sleep_midpoint'] -
    corrected_values['weekday_sleep_midpoint']).abs()

full_values = corrected_values.copy()
full_values 
```

```python
#Checking the final df shape and saving file to repo

print(full_values.shape)
full_values.to_csv(
    'data/02-processed/full_values.csv',
    index=False)

import os
os.listdir('data/02-processed')
```
